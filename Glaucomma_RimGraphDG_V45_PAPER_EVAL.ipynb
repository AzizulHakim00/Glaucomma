# RimGraph-DG V4.5 — Paper Evaluation Audit
CPU-only evaluator. It reads the existing `paper_run_v45` artifacts from Google Drive and does **not** modify training checkpoints.


```python
from google.colab import drive
drive.mount('/content/drive')

import subprocess, sys, urllib.request
from pathlib import Path

RUN_ROOT = Path('/content/drive/MyDrive/Glaucomma_RimGraphDG/paper_run_v45')
SCRIPT_URL = 'https://raw.githubusercontent.com/AzizulHakim00/Glaucomma/c579551cd72ac2424b2f73f292d05e42f8f6e9af/tools/rimgraph_v45_paper_eval.py'
SCRIPT = Path('/content/rimgraph_v45_paper_eval.py')

print('=== RIMGRAPH V4.5 PAPER EVALUATION LAUNCHER ===')
print('Run root:', RUN_ROOT)
if not RUN_ROOT.exists():
    raise FileNotFoundError(f'Run folder not found: {RUN_ROOT}')

urllib.request.urlretrieve(SCRIPT_URL, SCRIPT)
subprocess.run([sys.executable, '-m', 'py_compile', str(SCRIPT)], check=True)
print('[CHECK] evaluator downloaded and compiled successfully')

cmd = [
    sys.executable, str(SCRIPT), str(RUN_ROOT),
    '--targets', 'ORIGA',
    '--seeds', '2029',
    '--bootstrap', '500',
]
print('[RUN] ORIGA / seed 2029 readiness audit')
subprocess.run(cmd, check=True)

OUT = RUN_ROOT / 'paper_eval_v45'
print('\n✅ EVALUATION AUDIT FINISHED')
print('Results saved to:', OUT)
print('Readiness:', OUT / 'paper_readiness.json')
print('Completeness:', OUT / 'experiment_completeness.csv')
print('Metrics:', OUT / 'collected_metrics.csv')
print('Paired bootstrap:', OUT / 'paired_bootstrap_baseline_vs_rimgraph.csv')
```
